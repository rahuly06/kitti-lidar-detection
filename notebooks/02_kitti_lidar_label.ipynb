```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# loading the lidar point cloud data from the KITTI dataset
raw = np.fromfile("../data/KITTI/training/velodyne/000000.bin", dtype=np.float32)
points = raw.reshape(-1, 4)

# plotting the point cloud data
fig = plt.figure(figsize=(10, 10))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(points[:, 0], points[:, 1], points[:, 2], c=points[:, 2], cmap='viridis', s=1)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z') 
```

```python
# loading the KITTI dataset labels

with open("../data/KITTI/training/label_2/000000.txt") as f:
    lines = f.readlines()

print(len(lines))
print(lines[0])
```

```python
parts = lines[0].strip().split()

for i, value in enumerate(parts):
    print(f'{i}:', value)
```

```python
"""
#Values    Name      Description
----------------------------------------------------------------------------
   1    type         Object type (Car, Van, Truck, etc.)
   1    truncated    Float 0-1 (truncated ratio)
   1    occluded     Integer (0=visible, 1=partly occluded, 2=fully occluded)
   1    alpha        Observation angle (-pi..pi)
   4    bbox         2D bounding box (x1,y1,x2,y2) in pixels
   3    dimensions   3D dimensions (height, width, length) in meters
   3    location     3D location (x,y,z) in camera coordinates
   1    rotation_y   Rotation around Y-axis in camera coordinates
"""

object_type = parts[0]
dimensions = [parts[8], parts[9], parts[10]]
height = parts[8]
width = parts[9]
length = parts[10]
location = [parts[11], parts[12], parts[13]]
loc_x = parts[11]
loc_y = parts[12]
loc_z = parts[13]
rotation_y = parts[14]

print(f'The object type is: {object_type} \nThe dimensions are {dimensions} \nThe location is {location}')
```
